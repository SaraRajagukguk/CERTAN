```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
# Load dataset Iris dari scikit-learn
iris = load_iris()
```

Pada code ini digunakan untuk mempersiapkan library yang diperlukan dalam pengolahan dataset Iris. numpy digunakan untuk pengolahan data numerik, pandas digunakan untuk mengolah data dalam bentuk DataFrame, sedangkan load_iris digunakan untuk mengambil dataset Iris yang tersedia pada Scikit-Learn. Dataset Iris berisi data pengukuran bunga Iris yang nantinya digunakan sebagai data dalam proses klasifikasi menggunakan algoritma Naive Bayes.

```python
# Konversi data menjadi DataFrame Pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]
# Definisikan X (fitur) dan y (target/label)
X = df.drop(columns=["Species"])
y = df["Species"]
print(X)
```

Pada code ini digunakan untuk mengolah dataset Iris menjadi bentuk DataFrame, menambahkan label species, kemudian memisahkan data fitur dan target yang akan digunakan dalam proses klasifikasi. pd.DataFrame() digunakan untuk mengubah data Iris menjadi DataFrame dengan nama kolom yang berasal dari iris.feature_names. Selanjutnya, df["Species"] digunakan untuk menambahkan kolom species berdasarkan target pada dataset Iris. Setelah itu, X digunakan sebagai data fitur dengan menghapus kolom Species, sedangkan y digunakan sebagai target atau label yang berisi nama species. Terakhir, print(X) digunakan untuk menampilkan seluruh data fitur yang akan digunakan dalam proses klasifikasi.   

```python
y
```

Pada code y digunakan untuk menentukan target atau label yang akan diprediksi oleh model. y = df["Species"] mengambil seluruh data pada kolom Species, sehingga y berisi nama species bunga Iris seperti setosa, versicolor, dan virginica. Data pada y nantinya akan diubah menjadi bentuk angka menggunakan LabelEncoder agar dapat digunakan dalam proses klasifikasi

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

Pada code ini digunakan untuk mengubah label species yang sebelumnya berupa teks menjadi bentuk angka agar dapat digunakan dalam proses klasifikasi. LabelEncoder digunakan untuk membuat encoder, kemudian lab.fit_transform(y) digunakan untuk mempelajari dan mengubah nilai pada y menjadi angka. Hasilnya adalah setosa menjadi 0, versicolor menjadi 1, dan virginica menjadi 2. Setelah itu, y digunakan untuk menampilkan hasil perubahan label tersebut dalam bentuk array angka.

```python
from sklearn.model_selection import train_test_split
X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

Pada code ini digunakan untuk membagi dataset menjadi data training dan data testing yang akan digunakan dalam proses klasifikasi. train_test_split digunakan untuk melakukan pembagian data, kemudian X dibagi menjadi X_train sebagai data fitur untuk training dan X_test sebagai data fitur untuk testing. Begitu juga dengan y, yang dibagi menjadi y_train sebagai label training dan y_test sebagai label testing. test_size=0.2 berarti 20% data digunakan untuk testing dan 80% untuk training, sedangkan random_state=0 digunakan agar pembagian data tetap sama setiap kali kode dijalankan.

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

Pada code ini digunakan untuk mengimpor dan membuat model Gaussian Naïve Bayes yang digunakan untuk melakukan klasifikasi pada dataset Iris. GaussianNB() digunakan untuk membuat model Naive Bayes dengan distribusi Gaussian, sedangkan model.fit(X_train, y_train) digunakan untuk melatih model menggunakan data fitur X_train dan label y_train. Hasil model yang telah dilatih kemudian disimpan ke dalam variabel NB_model sehingga dapat digunakan untuk melakukan prediksi pada data testing maupun data baru.

```python
result = NB_model.predict(X_test)
result
```

Pada code ini digunakan untuk melakukan prediksi terhadap data testing menggunakan model Gaussian Naïve Bayes yang sebelumnya sudah dilatih. NB_model.predict(X_test) digunakan untuk memprediksi species berdasarkan data fitur pada X_test, kemudian hasil prediksi disimpan ke dalam variabel result. Setelah itu, result digunakan untuk menampilkan hasil prediksi dalam bentuk angka, yaitu 0 untuk setosa, 1 untuk versicolor, dan 2 untuk virginica

```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

Pada code ini digunakan untuk menghitung tingkat akurasi model dalam melakukan klasifikasi pada data testing. accuracy_score(y_test, result) digunakan untuk membandingkan label sebenarnya pada y_test dengan hasil prediksi pada result. Hasil akurasi kemudian dikalikan 100 agar ditampilkan dalam bentuk persentase. Berdasarkan hasil pada modul, model menghasilkan accuracy sebesar 96.66666666666667
