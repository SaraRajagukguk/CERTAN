```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
# Load dataset Iris dari scikit-learn
iris = load_iris()
```

Pada code numpy dan pandas digunakan untuk mengolah data, sedangkan load_iris digunakan untuk mengambil dataset Iris dari Scikit-Learn. Kemudian load_iris() digunakan untuk memuat dataset Iris ke dalam variabel iris.

```python
# Konversi data menjadi DataFrame Pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]
# Definisikan X (fitur) dan y (target/label)
X = df.drop(columns=["Species"])
y = df["Species"]
print(X)
```

Code pd.DataFrame() digunakan untuk menyusun data Iris menjadi tabel dengan nama kolom yang sesuai dengan fitur pada dataset. Kemudian df["Species"] digunakan untuk menambahkan jenis bunga berdasarkan target yang ada pada dataset. Setelah itu, X berisi empat fitur bunga yang digunakan sebagai data masukan, sedangkan y berisi Species yang menjadi target klasifikasi. Kemudian, print(X) digunakan untuk menampilkan data fitur yang akan digunakan dalam proses klasifikasi.

```python
y
```

y digunakan untuk menampilkan data pada variabel y yang berisi jenis atau species bunga Iris. Data ini menjadi target/label yang nantinya digunakan model untuk belajar dan menentukan species dari data baru.

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

LabelEncoder() digunakan untuk memberi kode angka pada nama species agar bisa diproses oleh model.lalu lab.fit_transform(y) mengubah label pada y menjadi angka, yaitu setosa = 0, versicolor = 1, dan virginica = 2. Kemudian y digunakan untuk melihat hasil label yang sudah diberi kode angka.

```python
from sklearn.model_selection import train_test_split
X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

Code train_test_split() digunakan untuk membagi data menjadi data training dan data testing. X_train dan y_train digunakan untuk melatih model, sedangkan X_test dan y_test digunakan untuk menguji hasil prediksi. test_size=0.2 berarti 20% data untuk testing dan 80% untuk training, sedangkan random_state=0 agar pembagian datanya tetap sama saat dijalankan kembali.

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

GaussianNB() digunakan untuk membuat model Gaussian Naïve Bayes yang sesuai untuk data Iris karena fiturnya berupa data numerik. Kemudian model.fit(X_train, y_train) digunakan untuk melatih model menggunakan data training, yaitu X_train sebagai fitur dan y_train sebagai target. Model yang sudah dilatih disimpan dalam NB_model untuk digunakan pada proses prediksi.

```python
result = NB_model.predict(X_test)
result
```

pada bagian NB_model.predict(X_test) digunakan untuk memprediksi jenis bunga berdasarkan data testing yang sudah disiapkan. Hasil prediksi tersebut disimpan dalam variabel result, kemudian result digunakan untuk melihat hasil klasifikasi dari data testing.

```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

Pada code ini, accuracy_score() digunakan untuk menghitung seberapa tepat hasil prediksi model dibandingkan dengan data sebenarnya. y_test berisi label sebenarnya, sedangkan result berisi hasil prediksi dari model. Hasilnya dikalikan 100 agar ditampilkan dalam bentuk persen, yaitu 96,67% pada modul ini.
