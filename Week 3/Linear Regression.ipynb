```python
import pandas as pd
df = pd.read_csv('Salary_Data.csv') 
print(df.head())
```

```python
X = df.iloc[:, :-1].values 
y = df.iloc[:, 1].values
```

```python
import matplotlib.pyplot as plt 
plt.scatter(X, y, color = "red") 
plt.xlabel('Years Experience')
plt.ylabel('Salary') 
plt.show()
```

```python
#bagi dataset jadi train dan test 80:20
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state=0)
                                                    # bagi dataset jadi train dan test 80:20
```

```python
# Fitting Simple Linear Regression to the Training set

from sklearn.linear_model import LinearRegression

regressor = LinearRegression()
regressor.fit(X_train, y_train)
```

```python
a = regressor.intercept_
b = regressor.coef_[0]

print("Estimated model intercept, a:", a)
print("Estimated model slope, b:", b)
print(f"ŷ = {a:.2f} + {b:.2f}x")
```

```python
# memprediksi hasil pada data uji (Test set) menggunakan model regresi yang tela 
y_pred = regressor.predict(X_test)
y_pred
```

```python
# Menghitung nilai R-Square, yang mengukur seberapa baik model regresi linear 
# # Semakin mendekati 1, semakin baik model

from sklearn.metrics import r2_score 
score=r2_score(y_test,y_pred)
score
```

```python
# Visualizing the Training set results
plt.scatter(X_train, y_train, color = 'red')
plt.plot(X_train, regressor.predict(X_train), color = 'blue')
```

```python
# Visualizing the Test set results
plt.scatter(X_test, y_test, color = 'red')
plt.plot(X_train, regressor.predict(X_train), color = 'blue') 
plt.title('Salary vs Experience (Test set)')
plt.xlabel('Years of Experience') 
plt.ylabel('Salary')
plt.show()
```

```python
#TUGAS
#NOMOR 1

import pandas as pd

df = pd.read_csv('rounded_hours_student_scores.csv')
print(df.head())
```

```python
print(df.columns)
```

```python
# Menentukan predictor (X) dan target (y)

X = df.iloc[:, :-1].values
y = df.iloc[:, 1].values

print(X[:5])
print(y[:5])
```

```python
# Visualisasi data

import matplotlib.pyplot as plt

plt.scatter(X, y, color='red')
plt.xlabel('Study Hours')
plt.ylabel('Score')
plt.title('Study Hours vs Score')
plt.show()
```

```python
# Bagi dataset menjadi train dan test 80:20

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=0
)
```

```python
# Fitting Simple Linear Regression to the Training set

from sklearn.linear_model import LinearRegression

regressor = LinearRegression()
regressor.fit(X_train, y_train)
```

```python
# Menghitung intercept dan slope

a = regressor.intercept_
b = regressor.coef_[0]

print("Estimated model intercept, a:", a)
print("Estimated model slope, b:", b)
print(f"ŷ = {a:.2f} + {b:.2f}x")
```

```python
# Melakukan prediksi 

y_pred = regressor.predict(X_test)

print(y_pred)
```

```python
# Menghitung nilai R-Square

from sklearn.metrics import r2_score
score = r2_score(y_test, y_pred)

print("R-Square:", score)
```

```python
# Visualizing the Training set results

plt.scatter(X_train, y_train, color='red')
plt.plot(X_train, regressor.predict(X_train), color='blue')
plt.title('Study Hours vs Score (Training set)')
plt.xlabel('Study Hours')
plt.ylabel('Score')
plt.show()
```

```python
# Visualisasi test set results

plt.scatter(X_test, y_test, color='red')
plt.plot(X_train, regressor.predict(X_train), color='blue')
plt.title('Study Hours vs Score (Test set)')
plt.xlabel('Study Hours')
plt.ylabel('Score')
plt.show()
```
